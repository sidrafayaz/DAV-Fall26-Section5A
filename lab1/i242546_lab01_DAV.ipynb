## Task 10 - Data Profiling Summary

### Dataset Overview

The dataset contains information about people including age, workclass, education, marital status, occupation, sex, hours worked per week and income. The dataset has 32561 rows and 15 columns.

### Data Quality Issues

The missing values were stored as "?" so they were not detected by isnull() at first. After replacing "?" with NaN, missing values were found in workclass, occupation and native.country.

There are 1836 missing values in workclass, 1843 missing values in occupation and 583 missing values in native.country.

There are 24 exact duplicate rows. When the income column is ignored, there are 25 duplicate records.

### Key Observations

The mean age is approximately 38.58 and the median age is 37. The mean hours-per-week is approximately 40.44 and the median is 40. The mean and median values are fairly close for both variables.

The most common occupation is Prof-specialty with 4140 records.

The sex distribution is approximately 66.92% male and 33.08% female.

The income distribution is approximately 75.92% <=50K and 24.08% >50K. Therefore, the income classes are imbalanced.

The education and education-num columns are consistent because every education level maps to only one education-num value.

### Cleaning Decisions

The missing categorical values in workclass, occupation and native.country were replaced with "Unknown". This was done instead of dropping the rows because these columns are categorical and dropping the missing records would reduce the size of the dataset.

There were no missing numeric values, so no numeric imputation was needed.

Exact duplicate rows were identified and can be removed because they represent repeated records. Duplicate records found when ignoring the income column were not treated as automatic duplicates because two people can have the same attributes but different income outcomes.

The education, marital.status and native.country columns were stripped using str.strip() to remove leading and trailing whitespace.
