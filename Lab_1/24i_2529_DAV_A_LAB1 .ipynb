```python
#import libraries
import pandas as pd
import numpy as np
import sqlite3

#load dataset
df = pd.read_csv('adult.csv')
```

```python
#Task-1:fix hidden missing values
df.head()
df.info()
print(df.isnull().sum())
df.replace('?', np.nan, inplace=True)
print(df.isnull().sum())
```

```python
#Task-2:Qunatify Meaningness
missing = pd.DataFrame({
    'Missing Count': df.isnull().sum(),
    'Missing Percentage': df.isnull().mean() * 100
}).sort_values('Missing Count', ascending=False)

print(missing)
print(df[['workclass', 'occupation', 'native.country']].isnull().sum())
```

```python
#Task-3 Handle missing values
df['workclass'] = df['workclass'].fillna('Unknown')
df['occupation'] = df['occupation'].fillna('Unknown')
df['native.country'] = df['native.country'].fillna('Unknown')

print(df.isnull().sum())
```

```python
#Task-4:Detect duplicates
print('Exact duplicates:', df.duplicated().sum())
print(df[df.duplicated(keep=False)])
print('Duplicates ignoring income:', df.duplicated(subset=df.columns.drop('income')).sum())
df.drop_duplicates(inplace=True)
print('Shape after removing duplicates:', df.shape)
```

```python
#Task-5:Fix Inconsistent Categorical Data
print(df['education'].unique())
print(df['marital.status'].unique())
print(df['native.country'].unique())

df['education'] = df['education'].str.strip()
df['marital.status'] = df['marital.status'].str.strip()
df['native.country'] = df['native.country'].str.strip()

print(df['education'].unique())
print(df['marital.status'].unique())
print(df['native.country'].unique())
```

```python
#Task 6:Descriptive Statistics
print(df.describe())
print(df.describe(include='object'))

print('Mean Age:', df['age'].mean())
print('Median Age:', df['age'].median())
print('Mean Hours per Week:', df['hours.per.week'].mean())
print('Median Hours per Week:', df['hours.per.week'].median())
```

```python
#Task 7:Frequency Analysis
print('Occupation:')
print(df['occupation'].value_counts())

print('Most Common Occupation:', df['occupation'].value_counts().idxmax())

print('Sex Percentage:')
print(df['sex'].value_counts(normalize=True) * 100)

print('Income Percentage:')
print(df['income'].value_counts(normalize=True) * 100)
```

```python
#Task 8:Education Consistency
education_check = df.groupby('education')['education.num'].unique()

print(education_check)

print('Inconsistencies:')
print(education_check[education_check.apply(len) > 1])
```
