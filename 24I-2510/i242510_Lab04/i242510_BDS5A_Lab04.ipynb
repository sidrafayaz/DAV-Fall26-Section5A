```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('store.csv', encoding='latin1')

df.info()

df
```

```python
# Task 1

# Historgram of sales, kde = True
plt.figure(figsize=(10, 6))
sns.histplot(df['Sales'], kde=True)
plt.title('Distribution of Sales')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.show()

# Skewness
skewness_value = df['Sales'].skew()
print(f"Skewness of Sales: {skewness_value:.2f}")

# Mean vs Median
mean_sales = df['Sales'].mean()
median_sales = df['Sales'].median()
print(f"Mean of Sales: {mean_sales:.2f}")
print(f"Median of Sales: {median_sales:.2f}")

# Positively Skewed as SKewness = 12.97 > 0
```

```python
# Task 2

# Create a figure with two subplots arranged side-by-side
fig, axes = plt.subplots(1, 2, figsize=(18, 7))

# Violin Plot - Profit by Category on the first subplot
sns.violinplot(x='Category', y='Profit', data=df, palette='muted', ax=axes[0])
axes[0].set_title('Violin Plot of Profit by Category')
axes[0].set_xlabel('Category')
axes[0].set_ylabel('Profit')
axes[0].grid(axis='y', linestyle='--', alpha=0.7)

# Strip Plot - Profit by Category on the second subplot
sns.stripplot(x='Category', y='Profit', data=df, jitter=True, alpha=0.5, ax=axes[1])
axes[1].set_title('Strip Plot of Profit by Category (Jittered)')
axes[1].set_xlabel('Category')
axes[1].set_ylabel('Profit')
axes[1].grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
# Even tho strip plot gives individual points, violin plot shows distirbution, peaks and density which is used to recognize category in a bimodal distribution
```

```python
# Task 3

# Scatter Plot : discount vs profit with Category as hue and Sales as size
plt.figure(figsize=(12, 8))
sns.scatterplot(x='Discount', y='Profit', hue='Category', size='Sales', sizes=(20, 400), data=df, alpha=0.7)
plt.title('Scatter Plot of Discount vs Profit by Category (Size by Sales)')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.legend(title='Category', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# Bigger circles (sales) when less discount is applied
```

```python
# Task 4

# Numerical Columns
numeric_cols = df.select_dtypes(include=np.number).columns
df_numeric = df[numeric_cols]

# Pearson correlation matrix
pearson_corr = df_numeric.corr(method='pearson')
print("\nPearson Correlation Matrix:")
print(pearson_corr)

# Spearman correlation matrix
spearman_corr = df_numeric.corr(method='spearman')
print("\nSpearman Correlation Matrix:")
print(spearman_corr)

# 4. Compare matrices and identify disagreement
diff_corr = abs(pearson_corr - spearman_corr)
np.fill_diagonal(diff_corr.values, 0) # Set diagonal to 0 as correlation with self is 1 for both

# Find the pair with the maximum difference
max_diff_val = diff_corr.max().max()
max_diff_row = diff_corr.stack().idxmax()[0]
max_diff_col = diff_corr.stack().idxmax()[1]

print(f"\nMaximum difference in correlation found between '{max_diff_row}' and '{max_diff_col}': {max_diff_val:.4f}")
print(f"Pearson correlation: {pearson_corr.loc[max_diff_row, max_diff_col]:.4f}")
print(f"Spearman correlation: {spearman_corr.loc[max_diff_row, max_diff_col]:.4f}")

# 5. Inspect their scatter plot
plt.figure(figsize=(8, 6))
sns.scatterplot(x=df_numeric[max_diff_row], y=df_numeric[max_diff_col])
plt.title(f'Scatter Plot of {max_diff_row} vs {max_diff_col}')
plt.xlabel(max_diff_row)
plt.ylabel(max_diff_col)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# 6. Explain the disagreement
print("\nExplanation for correlation disagreement:")
print(f"The pair '{max_diff_row}' and '{max_diff_col}' shows a noticeable difference between Pearson and Spearman correlations. This typically occurs when the relationship between variables is not strictly linear, or when outliers are present. Pearson correlation measures the linear relationship and is sensitive to outliers and non-normal distributions, while Spearman correlation measures the monotonic relationship (rank correlation) and is more robust to outliers and non-linear but monotonic associations. Examining the scatter plot helps to confirm the nature of this disagreement – for instance, if there's a strong non-linear trend or significant outliers pulling the Pearson coefficient away from the Spearman coefficient.")
```

```python
# Task 5

# Region x Category Table of Sales with margin as True
sales_pivot = pd.pivot_table(df, values='Sales', index='Region', columns='Category', aggfunc='sum', margins=True)
print("Region x Category Total Sales Pivot Table (with Margins):")
print(sales_pivot)

# Render it as a heatmap with a sequential palette
plt.figure(figsize=(10, 7))
sns.heatmap(sales_pivot, annot=True, fmt=".0f", cmap="Greens", linewidths=.5)
plt.title('Total Sales by Region and Category (with Margins)')
plt.xlabel('Category')
plt.ylabel('Region')
plt.tight_layout()
plt.show()

# East for Technology is the biggest contributor to revenue.
```

```python
# Task 6

# Setting Order Date to datetime format and setting it as index
df['Order Date'] = pd.to_datetime(df['Order Date'])
df_time_series = df.set_index('Order Date')

# Resample sales to monthly totals
monthly_sales = df_time_series['Sales'].resample('MS').sum()

# Calculate a 3-month rolling average
rolling_avg = monthly_sales.rolling(window=3).mean()

# Plot monthly sales and rolling average
plt.figure(figsize=(12, 6))
plt.plot(monthly_sales.index, monthly_sales.values, label='Monthly Sales', alpha=0.7)
plt.plot(rolling_avg.index, rolling_avg.values, label='3-Month Rolling Average', color='red', linewidth=2)
plt.title('Monthly Sales vs. 3-Month Rolling Average')
plt.xlabel('Date')
plt.ylabel('Total Sales')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# Describe what the rolling average reveals
print("\nThe rolling average smooths out short-term fluctuations, making the underlying trend in sales more apparent. This helps to distinguish genuine long-term patterns and cycles from random noise or seasonal spikes.")
```

```python
# Task 8

# 1. Raw order-level discount vs. profit scatter
plt.figure(figsize=(10, 6))

sns.scatterplot(x='Discount', y='Profit', data=df, s=20)
plt.title('Raw Discount vs. Profit Scatter Plot (with overplotting)')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()


# 2. Fix overplotting using hexbin
plt.figure(figsize=(10, 6))

plt.hexbin(x=df['Discount'],y=df['Profit'],gridsize=50,cmap='viridis',mincnt=1)
plt.colorbar(label='Count of Orders')
plt.title('Discount vs. Profit Hexbin Plot (Addressing Overplotting)')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.grid(True, linestyle='--', alpha=0.6)

# Justification
print(
    "Justification: Hexbin is chosen because the large number of "
    "data points causes heavy overplotting, and hexbin clearly "
    "shows areas of high data density."
)


# 3. Find the most profitable order and largest loss

most_profitable_order = df.loc[df['Profit'].idxmax()]
largest_loss_order = df.loc[df['Profit'].idxmin()]


# 4. Annotate the most profitable order

plt.annotate(
    f"Most Profitable\nProfit: ${most_profitable_order['Profit']:.2f}",

    # Actual point to annotate
    xy=(
        most_profitable_order['Discount'],
        most_profitable_order['Profit']
    ),

    # Where to place the text
    xytext=(20, 20),

    # xytext is measured in points relative to xy
    textcoords='offset points',

    # Draw arrow from text to point
    arrowprops=dict(arrowstyle='->'),

    # Text box
    bbox=dict(boxstyle='round,pad=0.5'),

    fontsize=9
)


# 5. Annotate the largest loss

plt.annotate(
    f"Largest Loss\nProfit: ${largest_loss_order['Profit']:.2f}",

    # Actual point to annotate
    xy=(
        largest_loss_order['Discount'],
        largest_loss_order['Profit']
    ),

    # Place text below/right of the point
    xytext=(20, -30),

    textcoords='offset points',

    arrowprops=dict(arrowstyle='->'),

    bbox=dict(boxstyle='round,pad=0.5'),

    fontsize=9
)


plt.tight_layout()
plt.show()
```
