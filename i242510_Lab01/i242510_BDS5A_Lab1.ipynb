```python
import pandas as pd
import numpy as np
import sqlite3
```

```python
df = pd.read_csv("adult.csv")
df
```

```python
#Task 1: Fix Hidden Missing Values
df.info()

#Before Replacing
df.isnull().sum()

#After Replacing
df.replace('?',np.nan, inplace = True)
df.isnull().sum()
```

```python
# Task 2: Quantify Meaningness
missing = pd.DataFrame({
    'Missing Count': df.isnull().sum(),
    'Missing Percentage': df.isnull().mean() * 100
})

missing = missing.sort_values( by='Missing Percentage', ascending=False)
missing

#Occupation Has the highest missing percentage
```

```python
# Task 3: Handle Missing Values:
# Reason: Non conventional Occupations are included here which is a different category of its own.
df[['occupation']] = df[['occupation']].fillna("Unknown")
df[['workclass']] = df[['workclass']].fillna("Unknown")

#Dropping because insignificant / too less.
df.dropna(subset=['native.country'], inplace=True)
```

```python
# Task 4: Detect Duplicates
df.duplicated().sum()
print("Duplicated Values: ", df.duplicated().sum())

#df.duplicated(keep=False)

#Dropping Duplicates because fnlwgt can almost be used as an unique identifier, if its the same it implies its repeated data.
df = df.drop_duplicates()
df.duplicated().sum()
print("Duplicated Values: ", df.duplicated().sum())
```

```python
# Task 5: Fix Inconsistent categorical Data:

print("education unique values:")
print(df['education'].unique())

print("\nmarital.status unique values:")
print(df['marital.status'].unique())

print("\nnative.country unique values:")
print(df['native.country'].unique())

# No Leading or Trailing whitespace Found
```

```python
# Task 6: Descriptive Statistics:

print("Numerical Columns Description: ")
print(df.describe())

print("Categorical Columns Description: ")
print(df.describe(include = 'object'))

print("\nAge Mean: ", df['age'].mean())
print("Age Median: ", df['age'].median())
# Mean Differs from Median, Rightly skewed as Mean > Median, More older people

print("\nHours_per_week Mean: ", df['hours.per.week'].mean())
print("Hours_per_week Median: ", df['hours.per.week'].median())
# Mean Almost Equal to median - Almost normal distribution
```

```python
# Task 7 : Frequency Analysis

# Most common occupation
occupation_counts = df['occupation'].value_counts()
print("Occupation value counts:")
print(occupation_counts)
print(f"\nMost common occupation: {occupation_counts.idxmax()} ({occupation_counts.max()} records)")

# Percentage distribution of sex
print()
sex = df['sex'].value_counts(normalize=True) * 100
print(sex.round(2))
#Imbalanced, Ratio - 2:1

# (c) Percentage distribution of income label
print()
income = df['income'].value_counts(normalize=True) * 100
print(income.round(2))
# Very Imbalanced - 4:1
```

```python
# Task 8: Cross-Check a Data Quality Assumption
df.groupby('education')['education.num'].unique()

# each education value maps to exactly one number, no inconsistencies found.
```

```python
# Task 9:
conn = sqlite3.connect('adult_income.db')
df.to_sql('adult_income', conn, if_exists='replace', index=False)

query = "SELECT * FROM adult_income WHERE age > 30;"
query_result = pd.read_sql_query(query, conn)

print(f"Filtered SQL query (age > 30) shape: {query_result.shape}")
```
