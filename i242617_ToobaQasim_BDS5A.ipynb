```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('adult_income.db')
df_csv = pd.read_csv('adult.csv')
df_csv.to_sql('adult_income', conn, if_exists='replace', index=False)
```

```python
query = "SELECT * FROM adult_income WHERE age > 30"
df_sql = pd.read_sql_query(query, conn)
```

```python
print("CSV shape:", df_csv.shape)
print("SQL (age>30) shape:", df_sql.shape)

conn.close()
```
