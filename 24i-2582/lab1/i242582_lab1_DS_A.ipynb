```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('adult_income.db')

raw_df = pd.read_csv('adult.csv')

raw_df.to_sql(
    'adult_income',
    conn,
    if_exists='replace',
    index=False
)
```

```python
query = """
SELECT *
FROM adult_income
WHERE age > 30;
"""

sql_df = pd.read_sql_query(query, conn)

print(sql_df.head())
print(sql_df.shape)
```

Dataset Overview

The dataset contains 32,561 rows and 15 columns and is used to predict whether a person's income is <=50K or >50K.

Data Quality Issues

Missing values were found in occupation (1,843), workclass (1,836), and native.country (583). There were 24 exact duplicate rows. No inconsistencies were found between education and education.num.

Key Observations

Prof-specialty was the most common occupation. The income classes were imbalanced: 75.92% <=50K and 24.08% >50K. Mean and median age were 38.58 and 37, while mean and median hours-per-week were 40.44 and 40.

Cleaning Decisions

Missing categorical values were replaced with "Unknown". Exact duplicates can be removed, while rows with identical features but different income labels were retained. Categorical whitespace was stripped.
