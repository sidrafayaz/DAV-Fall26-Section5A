```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt


df= sns.load_dataset("penguins")
```

```python
df.head(100)
```

```python
#Task 1
average = df.bill_length_mm.mean()
ax = sns.barplot(x='species', y= 'bill_length_mm',data=df ,estimator = 'mean')
ax.set(
  title = 'Average bill length by species',
  xlabel = 'Species',
  ylabel = 'Average bill length (mm)'

);
```

```python
#Task 2
ax = sns.countplot(x='island',data =df)
ax.set(
  title = 'Number of penguins per island',
  xlabel = 'Island',
  ylabel = 'Number of penguins'


);
```

```python
#Task 3
ax = sns.histplot(x='bill_depth_mm',data=df,kde = True)
ax.set(
  title = 'Distribution of Bill Depth',
  xlabel = 'Bill Depth (mm)',
  ylabel = 'Frequency');
```

```python
#Task4
ax = sns.boxplot(y='flipper_length_mm',data=df)
ax.set(
  title = 'Distribution of Flipper Length',
  xlabel = 'Species',
  ylabel = 'Flipper Length (mm)');
```

```python
#Task5
Q1=df['flipper_length_mm'].quantile(0.25)
Q3=df['flipper_length_mm'].quantile(0.75)
IQR=Q3-Q1
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
outliers=df[(df['flipper_length_mm']<lower_bound) | (df['flipper_length_mm']>upper_bound)]
print(len(outliers))
```

```python
#Task 6
from scipy import stats
df['zscore'] = stats.zscore(df['flipper_length_mm'])
outliers_z = df[df['zscore'].abs() > 3]
print(len(outliers_z))
```

```python
#Task 7
sns.scatterplot(x='bill_length_mm',y='bill_depth_mm',data=df,hue='species')


```

```python
#Task 9

sns.countplot(data=df,x='species')
plt.show()

ax=sns.countplot(data=df,x='species')
ax.set_ylim(bottom=50)
```

```python
#Task8
avg_depth = df.groupby('bill_length_mm')['bill_depth_mm'].mean()
ax = sns.lineplot(x=avg_depth.index, y=avg_depth.values)

ax.set(
    xlabel='Bill Length (mm)',
    ylabel='Average Bill Depth (mm)',
    title='Average Bill Depth Across Bill Length'
);
```

```python
plt.fill_between(avg_depth.index, avg_depth.values)
ax.set_ylim(bottom=15.5)
```
