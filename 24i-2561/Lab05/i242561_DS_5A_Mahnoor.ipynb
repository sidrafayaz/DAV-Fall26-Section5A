```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed(42)
df=pd.DataFrame({
    'Date': pd.date_range(start='2023-01-01', periods=100, freq='D'),
    'Country': np.random.choice(['USA', 'Canada', 'UK', 'Germany', 'France'], 100),
    'Segment': np.random.choice(['Government', 'Midmarket', 'Channel Partners'], 100),
    'Product': np.random.choice(['Carretera', 'Montana', 'Paseo'], 100),
    'Sales': np.random.uniform(10000, 100000, 100),
    'Profit': np.random.uniform(1000, 20000, 100)
})
```

## Task: **01**

```python
plt.figure(figsize=(8, 4))
sales_by_country = df.groupby('Country')['Sales'].sum().reset_index()
sns.barplot(data=sales_by_country, x='Country', y='Sales', palette='Blues_d')
plt.title('Task 1: Total Sales by Country')
plt.show()
```

## Task: **02**

```python
country_summary = df.groupby('Country')[['Sales', 'Profit']].sum().reset_index()
country_summary.plot(x='Country', y=['Sales', 'Profit'], kind='bar', figsize=(9, 5))
plt.title('Task 2: Sales vs Profit by Country')
plt.ylabel('Amount ($)')
plt.xticks(rotation=0)
plt.show()
```

## Task: **03**

```python
plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x='Sales', y='Profit', hue='Segment', style='Country', s=100)
plt.title('Task 3: Sales vs Profit grouped by Segment (Legend)')
plt.show()
```

## Task: **04**

```python
g = sns.FacetGrid(df, col="Segment", height=4, aspect=1.2)
g.map_dataframe(sns.barplot, x="Country", y="Sales", errorbar=None, palette="Set2")
g.fig.suptitle('Task 4: Sales by Country Split by Segment (Small Multiples)', y=1.05)
plt.show()
```
