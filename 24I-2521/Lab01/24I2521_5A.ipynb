### Task 1

```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('/content/sample_data/adult.csv')

# Initial inspection
print("First 5 rows:")
display(df.head())

print("\nDataset information:")
df.info()

# Check missing values before fixing hidden values
print("\nMissing values before replacing '?':")
print(df.isnull().sum())

# Replace hidden missing values with NaN
df.replace('?', np.nan, inplace=True)

# Check missing values after replacement
print("\nMissing values after replacing '?':")
print(df.isnull().sum())
```

### Task 2

```python
# Count missing values per column
missing_count = df.isnull().sum()

# Calculate missing percentage per column
missing_percentage = df.isnull().mean() * 100

# Create a summary table
missing_summary = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing Percentage': missing_percentage
})

# Sort by missing count in descending order
missing_summary = missing_summary.sort_values(
    by='Missing Count',
    ascending=False
)

display(missing_summary)
```

### Task 3

```python
# Check missing values before handling them
print("Missing values before handling:")
print(df.isnull().sum())

# Handle missing categorical values
df['workclass'] = df['workclass'].fillna('Unknown')
df['occupation'] = df['occupation'].fillna('Unknown')
df['native.country'] = df['native.country'].fillna('Unknown')

# Check missing values after handling
print("\nMissing values after handling:")
print(df.isnull().sum())

# Display the number of Unknown values introduced
print("\nUnknown values after imputation:")
print("workclass:", (df['workclass'] == 'Unknown').sum())
print("occupation:", (df['occupation'] == 'Unknown').sum())
print("native.country:", (df['native.country'] == 'Unknown').sum())
```

### Task 4

```python
# Check for exact duplicate rows
exact_duplicates = df.duplicated().sum()

print("Number of exact duplicate rows:")
print(exact_duplicates)

# Check for duplicates while ignoring the income target/label column
columns_without_target = df.columns.drop('income')

duplicates_without_target = df.duplicated(
    subset=columns_without_target
).sum()

print("\nNumber of duplicate rows ignoring the income label:")
print(duplicates_without_target)

# Display duplicate rows ignoring the income label
print("\nDuplicate rows ignoring the income label:")
display(
    df[df.duplicated(subset=columns_without_target, keep=False)]
    .sort_values(by=list(columns_without_target))
)
```

### Task 5

```python
# Display unique values before cleaning
print("Unique values in education:")
print(df['education'].unique())

print("\nUnique values in marital.status:")
print(df['marital.status'].unique())

print("\nUnique values in native.country:")
print(df['native.country'].unique())

# Remove leading and trailing whitespace from categorical columns
df['education'] = df['education'].str.strip()
df['marital.status'] = df['marital.status'].str.strip()
df['native.country'] = df['native.country'].str.strip()

# Display unique values after cleaning
print("\nUnique values in education after stripping:")
print(df['education'].unique())

print("\nUnique values in marital.status after stripping:")
print(df['marital.status'].unique())

print("\nUnique values in native.country after stripping:")
print(df['native.country'].unique())
```

### Task 6

```python
# Descriptive statistics for numeric columns
print("Descriptive Statistics for Numeric Columns:")
display(df.describe())

# Descriptive statistics for categorical columns
print("\nDescriptive Statistics for Categorical Columns:")
display(df.describe(include='object'))

# Mean and median for age
age_mean = df['age'].mean()
age_median = df['age'].median()

# Mean and median for hours-per-week
hours_mean = df['hours.per.week'].mean()
hours_median = df['hours.per.week'].median()

print("\nAge:")
print("Mean:", age_mean)
print("Median:", age_median)

print("\nHours per Week:")
print("Mean:", hours_mean)
print("Median:", hours_median)

# Compare mean and median
print("\nComparison:")
if abs(age_mean - age_median) < 1:
    print("Mean and median age are very close.")
else:
    print("Mean and median age differ meaningfully.")

if abs(hours_mean - hours_median) < 1:
    print("Mean and median hours-per-week are very close.")
else:
    print("Mean and median hours-per-week differ meaningfully.")
```

### Task 7

```python
# Most common occupation
occupation_counts = df['occupation'].value_counts()

print("Occupation Frequency:")
display(occupation_counts)

print("\nMost Common Occupation:")
print(occupation_counts.idxmax())
print("Count:", occupation_counts.max())

# Percentage distribution of sex
sex_percentage = df['sex'].value_counts(normalize=True) * 100

print("\nPercentage Distribution of Sex:")
display(sex_percentage)

# Percentage distribution of income
income_percentage = df['income'].value_counts(normalize=True) * 100

print("\nPercentage Distribution of Income:")
display(income_percentage)

# Check whether income classes are balanced or imbalanced
income_difference = abs(income_percentage.iloc[0] - income_percentage.iloc[1])

print("\nIncome Class Balance:")
if income_difference < 10:
    print("The income classes are relatively balanced.")
else:
    print("The income classes are imbalanced.")
```

### Task 8

```python
# Check the mapping between education and education.num
education_mapping = df.groupby('education')['education.num'].unique()

print("Education to education-num mapping:")
display(education_mapping)

# Identify education levels mapped to more than one number
inconsistencies = education_mapping[
    education_mapping.apply(lambda x: len(x) > 1)
]

print("\nInconsistent education mappings:")
if len(inconsistencies) == 0:
    print("No inconsistencies found.")
else:
    display(inconsistencies)

# Final result
print("\nResult:")
if len(inconsistencies) == 0:
    print("Education and education-num are consistent.")
else:
    print("Education and education-num contain inconsistencies.")
```

### Task 9

```python
import sqlite3
import pandas as pd

# Load the original CSV file
raw_df = pd.read_csv('/content/sample_data/adult.csv')

# Create a local SQLite database
conn = sqlite3.connect('/content/adult_income.db')

# Load the raw CSV data into a SQLite table
raw_df.to_sql(
    'adult_income',
    conn,
    if_exists='replace',
    index=False
)

print("Raw CSV loaded successfully into SQLite table: adult_income")

# Run SQL query
query = """
SELECT *
FROM adult_income
WHERE age > 30;
"""

# Load SQL query result into a Pandas DataFrame
sql_df = pd.read_sql_query(query, conn)

print("\nSQL Query:")
print(query)

print("First 5 rows extracted using SQL:")
display(sql_df.head())

print("\nShape of original CSV:")
print(raw_df.shape)

print("\nShape of SQL extracted DataFrame:")
print(sql_df.shape)

print("\nNumber of rows with age > 30:")
print(len(sql_df))

# Verify the extraction
print("\nSQL extraction completed successfully.")
```
